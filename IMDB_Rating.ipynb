```python
# import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# import Dataset
from google.colab import files
uploaded = files.upload()
```

```python
# import dataset
data = pd.read_csv('IMDB Dataset.csv')

data.head()

# Check Null Values
data.isnull().sum()
# No null values

# Plot Histogram fr feature Runtime

plt.hist(data['Runtime'])
plt.ylabel("Count of movies")
plt.xlabel("Runtime")

# PLot Histogram for featuer IMDB_Rating
plt.hist(data['IMDB_Rating'])
plt.ylabel("Count of movies")
plt.xlabel("IMDB Rating")

# Define parameter bins_runtime for feature Runtime and plot histogram or it
data['Runtime'].unique()

bins_time = np.arange(80, 230, 10)
plt.hist(data['Runtime'], edgecolor="black", bins=bins_time, color='g')
plt.ylabel("Count of movies")
plt.xlabel("Runtime")

# DEfine parameter bins_rating for feature IMDB_Rating and plot histogramfor it
data['IMDB_Rating'].unique()
bins_rating = np.arange(8, 10, 0.20)
plt.hist(data['IMDB_Rating'], edgecolor="Yellow", bins=bins_rating, color='Cyan')
plt.ylabel("Count of movies")
plt.xlabel("IMDB Rating")
plt.xticks(bins_rating)
```
